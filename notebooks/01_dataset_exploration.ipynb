```python

# Check whether each store-product combination
# has repeated observations over time.

print("Dataset columns:")
print(df.columns.tolist())

print("\nNumber of unique dates:", df["Date"].nunique())
print("Number of stores:", df["Store ID"].nunique())
print("Number of products:", df["Product ID"].nunique())

print("\nRepeated store-product-date records:")
print(
    df.duplicated(
        subset=["Date", "Store ID", "Product ID"]
    ).sum()
)

print("\nPrice summary:")
print(df["Price"].describe())

print("\nPrice variation by store and product:")
print(
    df.groupby(["Store ID", "Product ID"])["Price"]
    .nunique()
    .describe()
)
```

```python

print("Shape:", sales.shape)

print("\nColumns:")
print(sales.columns.tolist())

print("\nDuplicate rows:", sales.duplicated().sum())

print("\nDuplicate date-store-product records:",
      sales.duplicated(
          subset=["Date", "Store ID", "Product ID"]
      ).sum())

print("\nFirst 5 rows:")
display(sales.head())
```

```python
print("Dataset shape:", sales.shape)

print("\nAll columns:")
print(sales.columns.tolist())

print("\nFirst 5 rows:")
display(sales.head())

print("\nNumeric column summary:")
display(sales.describe().T)
```

```python
print("Rows and columns:", sales.shape)

print("\nDuplicate rows:", sales.duplicated().sum())

print("\nDate range:")
print(sales["Date"].min(), "to", sales["Date"].max())

print("\nUnique stores:", sales["Store ID"].nunique())
print("Unique products:", sales["Product ID"].nunique())

print("\nRows per date:")
print(sales.groupby("Date").size().describe())
```

```python
print("Price and Units Sold correlation:")
print(sales["Price"].corr(sales["Units Sold"]))

print("\nDiscount and Units Sold correlation:")
print(sales["Discount"].corr(sales["Units Sold"]))

print("\nAverage units sold by promotion:")
display(sales.groupby("Promotion")["Units Sold"].agg(["mean", "median", "count"]))

print("\nAverage units sold by discount level:")
display(sales.groupby("Discount")["Units Sold"].mean().sort_index())
```

```python
features = [
    "Price",
    "Discount",
    "Promotion",
    "Competitor Pricing",
    "Demand",
    "Units Sold"
]

display(sales[features].corr().round(3))
```
